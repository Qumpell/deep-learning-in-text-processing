```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
tasks = [
    "If John has 3 apples and buys 5 more, how many apples does he have in total?",
    "Anna has 11 dollars, spends 3 dollars, and then receives 5 dollars. How much money does she have at the end?",
    "A box contains 4 red balls and 6 blue balls. How many balls are in the box?",
    "If one pen costs 2 dollars, how much do 4 pens cost?"
]
```

```python
def generate_answer(prompt, max_tokens=64):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(**inputs, max_new_tokens=max_tokens)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)


answers_no_reasoning = {}
for task in tasks:
    prompt = f"Answer the question: {task}"
    answers_no_reasoning[task] = generate_answer(prompt)

answers_with_reasoning = {}
for task in tasks:
    prompt = f"Solve the problem step by step and give the final answer: {task}"
    answers_with_reasoning[task] = generate_answer(prompt)
```

```python
print("=== WITHOUT STEP-BY-STEP REASONING ===")
for k, v in answers_no_reasoning.items():
    print(f"\nTask: {k}\nAnswer: {v}")

print("\n=== WITH STEP-BY-STEP REASONING ===")
for k, v in answers_with_reasoning.items():
    print(f"\nTask: {k}\nAnswer: {v}")
```

## Analysis of Model Responses

### 1️ Correctness
- Without step-by-step reasoning: all answers are incorrect, ignores arithmetic operations, gives very short/partial answers.
- With step-by-step reasoning: also incorrect in these examples, often introduces extra or nonsensical steps.  
- Note: With a larger, more capable model, step-by-step reasoning would likely improve correctness, especially for multi-step tasks.

### 2️ Structure
- Without reasoning: very short, numeric only, hard to follow.
- With reasoning: longer, shows intermediate steps, more interpretable.

### 3️ When step-by-step reasoning helps
- Multi-step arithmetic or logic problems.
- Problems requiring understanding of context or sequence of actions.

### 4️ When step-by-step reasoning can mislead
- Very simple arithmetic: model may introduce extra steps and errors

### 5️ Summary
- Step-by-step reasoning improves interpretability and sometimes correctness.  
- For simple calculations, it can introduce errors; always verify the final answer.  
- Best used for complex, multi-step tasks rather than trivial arithmetic.
