```python
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(0)

batch = 64
seq_len = 1024    
model_dim = 768
d_k = 64
X = np.random.randn(batch, seq_len, model_dim) 
print("X shape:", X.shape, "\nX:\n", X, "\n")
```

```python
Wq = np.random.randn(model_dim, d_k) *  (1 / np.sqrt(model_dim))
Wk = np.random.randn(model_dim, d_k)  * (1 / np.sqrt(model_dim))
Wv = np.random.randn(model_dim, d_k)  * (1 / np.sqrt(model_dim))

print("Wq:\n", Wq, "\n")
print("Wk:\n", Wk, "\n")
print("Wv:\n", Wv, "\n")
```

```python
Q = X @ Wq  
K = X @ Wk
V = X @ Wv

print("Q:\n", Q, "\n")
print("K:\n", K, "\n")
print("V:\n", V, "\n")
```

```python
scores = np.zeros((batch, seq_len, seq_len))
for b in range(batch):
    scores[b] = Q[b] @ K[b].T

print("Scores (QK^T):\n", scores, "\n")
```

```python
scaled_scores = scores / np.sqrt(d_k)
print("Scaled scores:\n", scaled_scores, "\n")
```

```python
def stable_softmax(x):
    x = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(x)
    return exp / np.sum(exp, axis=-1, keepdims=True)
print("Max scaled score:", np.max(scaled_scores))
print("Min scaled score:", np.min(scaled_scores))
attn_probs = stable_softmax(scaled_scores)
print("Attention weights:\n", attn_probs, "\n")
```

```python
print("Row sums:\n", np.sum(attn_probs, axis=-1), "\n")
```
