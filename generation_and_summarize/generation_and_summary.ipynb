```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
prompt = "In the future, artificial intelligence will"
inputs = tokenizer(prompt, return_tensors="pt")
```

```python
greedy = model.generate(**inputs, max_length=50)
print("Greedy:\n", tokenizer.decode(greedy[0], skip_special_tokens=True))
```

```python
beam = model.generate(
    **inputs,
    max_length=50,
    num_beams=5,
    early_stopping=True
)
print("Beam search:\n", tokenizer.decode(beam[0], skip_special_tokens=True))
```

```python
top_k = model.generate(
    **inputs,
    max_length=50,
    do_sample=True,
    top_k=50
)
print("Top-k:\n", tokenizer.decode(top_k[0], skip_special_tokens=True))
```

```python
top_p = model.generate(
    **inputs,
    max_length=50,
    do_sample=True,
    top_p=0.9
)
print("Top-p:\n", tokenizer.decode(top_p[0], skip_special_tokens=True))
```

```python
texts = {
    "Greedy": tokenizer.decode(greedy[0], skip_special_tokens=True),
    "Beam search": tokenizer.decode(beam[0], skip_special_tokens=True),
    "Top-k": tokenizer.decode(top_k[0], skip_special_tokens=True),
    "Top-p": tokenizer.decode(top_p[0], skip_special_tokens=True)
}

for name, txt in texts.items():
    print("\n=== {} ===".format(name))
    print(f"Length: {len(txt.split())} words")
    print(txt)
```

Greedy → spójne, ale mało kreatywne, czasem zapętla się.

Beam search → bardziej spójne, logiczne, ale mało różnorodne.

Top-k → bardziej kreatywne, czasem mniej sensowne.

Top-p → najlepszy balans kreatywności i sensowności.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

sum_model_name = "t5-small"

sum_tokenizer = AutoTokenizer.from_pretrained(sum_model_name)
sum_model = AutoModelForSeq2SeqLM.from_pretrained(sum_model_name)
```

```python
text = """
On November 29th, Lachlan Davidson reported a security vulnerability in React that allows unauthenticated remote code execution by exploiting a flaw in how React decodes payloads sent to React Server Function endpoints.
Even if your app does not implement any React Server Function endpoints it may still be vulnerable if your app supports React Server Components.
This vulnerability was disclosed as CVE-2025-55182 and is rated CVSS 10.0.
The vulnerability is present in versions 19.0, 19.1.0, 19.1.1, and 19.2.0 of:
react-server-dom-webpack
react-server-dom-parcel
react-server-dom-turbopack
Immediate Action Required 
A fix was introduced in versions 19.0.1, 19.1.2, and 19.2.1. If you are using any of the above packages please upgrade to any of the fixed versions immediately.
If your app’s React code does not use a server, your app is not affected by this vulnerability. If your app does not use a framework, bundler, or bundler plugin that supports React Server Components, your app is not affected by this vulnerability.
Affected frameworks and bundlers 
Some React frameworks and bundlers depended on, had peer dependencies for, or included the vulnerable React packages. The following React frameworks & bundlers are affected: next, react-router, waku, @parcel/rsc, @vitejs/plugin-rsc, and rwsdk.
We will update this post with upgrade instructions on how to upgrade as they become available.
Hosting Provider Mitigations 
We have worked with a number of hosting providers to apply temporary mitigations.
You should not depend on these to secure your app, and still update immediately.
Vulnerability overview 
React Server Functions allow a client to call a function on a server. React provides integration points and tools that frameworks and bundlers use to help React code run on both the client and the server. React translates requests on the client into HTTP requests which are forwarded to a server. On the server, React translates the HTTP request into a function call and returns the needed data to the client.

An unauthenticated attacker could craft a malicious HTTP request to any Server Function endpoint that, when deserialized by React, achieves remote code execution on the server. Further details of the vulnerability will be provided after the rollout of the fix is complete.
"""
#source: https://react.dev/blog/2025/12/03/critical-security-vulnerability-in-react-server-components
```

```python
inputs = sum_tokenizer("summarize: " + text, return_tensors="pt", max_length=512, truncation=True)

summary_ids = sum_model.generate(
    **inputs,
    num_beams=4,
    max_length=120,
    early_stopping=True
)

summary = sum_tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("SUMMARY:\n", summary)
```

```python
import evaluate

rouge = evaluate.load("rouge")

scores = rouge.compute(
    predictions=[summary],
    references=[text]
)

print(scores)
```

### Summary Quality Evaluation

The generated summary captures the most important point from the original text — that several react-server-dom packages contain a critical vulnerability and that fixed versions are available. It’s factually correct and doesn’t introduce distortions.

However, the summary is very brief and leaves out many relevant details, such as the CVSS score, the risk of remote code execution, which frameworks are affected, and how the vulnerability works. This is reflected in the moderate ROUGE scores (~0.22), which show only partial overlap with the source text.

Overall, the summary is accurate but oversimplified. It conveys the essential message but lacks context that would help readers understand the severity and impact of the vulnerability.
